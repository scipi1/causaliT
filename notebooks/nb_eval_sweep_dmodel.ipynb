```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from os.path import join
```

# Uniform Sampling

```python
exp_path = "../experiments/noise_aware_single/scm1/euler/sweep_d_model_60863422/eval/eval_d_model_sweep"
df = pd.read_csv(join(exp_path, "files/sweep_summary.csv"))
df["d_model_hue"] = df["d_model"].astype(str)
df["shd_tot"] = df["shd_cross"] + df["shd_self"]
df['shd_tot_norm_by_group'] = df['shd_tot'] / df.groupby('d_model')['shd_tot'].transform('max')
df['test_loss_centered'] = (df['test_loss'] - df.groupby('d_model')['test_loss'].transform('min')) #/ df.groupby('d_model')['test_loss'].transform('max')
df['test_loss_norm_by_group'] = df['test_loss_centered'] / df.groupby('d_model')['test_loss_centered'].transform('max')

fig = plt.figure()
sns.boxplot(df, x="d_model_hue", y="test_loss")

fig = plt.figure()
sns.boxplot(df, x="d_model_hue", y="shd_tot")

fig = plt.figure()
sns.scatterplot(df, x="test_loss", y="shd_cross", label="cross")
sns.scatterplot(df, x="test_loss", y="shd_self", label="self")
sns.scatterplot(df, x="test_loss", y="shd_tot", label="tot")
plt.yscale("log")
plt.ylabel("SHD")
plt.legend()

fig = plt.figure()
sns.scatterplot(df, x="test_loss_norm_by_group", y="test_hsic", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="shd_tot", y="test_hsic", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")
```
