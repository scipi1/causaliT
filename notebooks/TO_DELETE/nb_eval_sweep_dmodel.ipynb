```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from os.path import join
```

# Uniform Sampling

```python
exp_path = "../experiments/noise_aware_single/scm1/euler/sweep_d_model_60863422/eval/eval_d_model_sweep"
df = pd.read_csv(join(exp_path, "files/sweep_summary.csv"))
df["d_model_hue"] = df["d_model"].astype(str)
df["shd_tot"] = df["shd_cross"] + df["shd_self"]
df['shd_tot_norm_by_group'] = df['shd_tot'] / df.groupby('d_model')['shd_tot'].transform('max')
df['test_loss_centered'] = (df['test_loss'] - df.groupby('d_model')['test_loss'].transform('min')) #/ df.groupby('d_model')['test_loss'].transform('max')
df['test_loss_norm_by_group'] = df['test_loss_centered'] / df.groupby('d_model')['test_loss_centered'].transform('max')
print(df.columns)

fig = plt.figure()
sns.boxplot(df, x="d_model_hue", y="test_loss")

fig = plt.figure()
sns.boxplot(df, x="d_model_hue", y="shd_tot")

fig = plt.figure()
sns.scatterplot(df, x="test_loss", y="shd_cross", hue="d_model_hue")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="test_loss", y="shd_self", hue="d_model_hue")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="test_loss", y="shd_tot", hue="d_model_hue")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="test_loss", y="test_hsic", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="test_loss_norm_by_group", y="test_hsic", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="shd_tot", y="test_hsic", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="shd_tot", y="val_hsic_mean_over_epochs", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="shd_cross", y="train_hsic_mean_over_epochs", hue="d_model_hue")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="shd_self", y="train_hsic_mean_over_epochs", hue="d_model_hue")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="shd_tot", y="train_hsic_mean_over_epochs", hue="d_model_hue")
plt.yscale("log")
plt.title("Discrete $S$ sampling")

fig = plt.figure()
sns.scatterplot(df, x="shd_cross", y="val_hsic_final", hue="d_model_hue", style="seed")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="shd_self", y="val_hsic_final", hue="d_model_hue", style="seed")
plt.yscale("log")

fig = plt.figure()
sns.scatterplot(df, x="shd_tot", y="val_hsic_final", hue="d_model_hue", style="seed")
plt.yscale("log")
plt.title("Discrete $S$ sampling")
```

```python
df['ate_mean_abs_error_std'] = df.groupby(["d_model", "seed"])["ate_mean_abs_error"].transform('std')
print(df['ate_mean_abs_error_std'].min())
print(df['ate_mean_abs_error_std'].max())
print(df['ate_mean_abs_error_std'].mean())
thres = 0.2
print(f"{sum(df['ate_mean_abs_error_std'].unique() > thres)} are available with the selected threshold")
```

```python
data = df[df['ate_mean_abs_error_std']>thres]

for d_model in data["d_model"].unique():
    fig = plt.figure()
    data_d_model = data.set_index(["d_model"]).loc[d_model].reset_index()
    for seed in data_d_model["seed"].unique():
        sns.regplot(data.set_index(["d_model","seed"]).loc[d_model].loc[seed], x="ate_mean_abs_error", y="val_hsic_final")
    plt.title(d_model)
```
